## KPI Function
I have created a reusable function (which is a set of instructions that perform a specific task) so that we don't repeat the same calculation. 

```python
def calculate_efficiency(actual_output, target_output):
    return (actual_output / target_output) * 100

def get_operational_status(efficiency):
    if efficiency < 70:
        return "Critical"
    elif efficiency < 90:
        return "Warning"
    return "Normal"
```

## Data Model
Each operational site is represented as a dictionary (a dictionary is a simple way of organizing related information) with its name, actual output, and target output.

```python
sites = [
    {"name": "Fuel Depot A", "actual_output": 650, "target_output": 1000},
    {"name": "Fuel Depot B", "actual_output": 820, "target_output": 1000},
    {"name": "Fuel Depot C", "actual_output": 960, "target_output": 1000},
]
```

## Execution
We loop through each site, calculate efficiency, determine the operational status, and print a simple report.

```python
for site in sites:
    efficiency = calculate_efficiency(site["actual_output"], site["target_output"])
    status = get_operational_status(efficiency)
    print(f"Site: {site['name']}")
    print(f"  Actual Output : {site['actual_output']}")
    print(f"  Target Output : {site['target_output']}")
    print(f"  Efficiency    : {efficiency:.2f}%")
    print(f"  Status        : {status}")
    print("-" * 40)
```
